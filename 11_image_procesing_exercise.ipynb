* Ambil gambar **objek dengan background bertekstur** (misalnya kain bermotif, jerami, atau batu).

````python
Filter Blur (Spatial Low-Pass Filter)
Filter Blur (seperti Gaussian Blur atau Median Blur) bekerja dengan cara meratakan nilai sebuah piksel dengan piksel-piksel di tetangga sekitarnya 
menggunakan kernel matriks tertentu.

Pengaruh terhadap Detail:
Filter ini bertindak sebagai Low-Pass Filter, artinya filter ini meloloskan komponen frekuensi rendah (area halus) dan membuang frekuensi 
    tinggi (detail tajam). Akibatnya, detail kecil, garis tajam, dan tekstur kasar pada gambar akan melunak dan menjadi kabur (smooth).

Dampak pada Noise:
Sangat efektif untuk mereduksi noise atau bintik-bintik kecil pada gambar, namun mengorbankan ketajaman batas tepi objek.
Filter Sharpening (Spatial High-Pass Filter / Unsharp Masking)
    
Filter Sharpening 
bekerja dengan cara memperkuat perbedaan kontras pada gradien tepi. Teknik ini sering kali mengurangi gambar asli dengan versi blur-nya untuk 
mendapatkan komponen detail high-frequency, lalu menambahkan kembali komponen tersebut ke gambar asli.

Pengaruh terhadap Detail:
Filter ini bertindak sebagai High-Pass Filter yang menonjolkan komponen frekuensi tinggi (tepi dan detail mikro). Tekstur kasar pada gambar 
(seperti urat batu atau serat jerami) akan terlihat jauh lebih kontras, tegas, dan menonjol.

Dampak pada Noise:
Sisi negatif dari filter ini adalah memicu munculnya artifact atau penajaman berlebih (oversharpening). Jika gambar asli memiliki noise atau
bintik halus, filter ini justru akan memperjelas dan memperburuk noise tersebut sehingga gambar terlihat kotor (grainy).
erbandingan Efek Filter Blur dan Sharpening dalam Memengaruhi Detail Gambar
Secara mendasar, filter blur dan filter sharpening bekerja secara bertolak belakang dalam memanipulasi frekuensi spasial dan detail dari sebuah gambar.
  
Perbandingan Efek Filter Blur dan Sharpening dalam Memengaruhi Detail Gambar
Filter blur bertindak sebagai Low-Pass Filter yang bekerja dengan cara meratakan nilai intensitas piksel dengan area di sekitarnya. 
Proses ini secara efektif membuang komponen frekuensi tinggi, yang mengakibatkan hilangnya detail-detail mikro, pelunakan tekstur
latar belakang yang kasar (seperti serat kain atau pori-pori batu), serta membuat batas tepi objek menjadi kabur dan halus (soft edges).
Sisi positifnya, filter blur sangat berguna untuk mereduksi noise atau bintik-bintik acak yang mengganggu visual gambar.

Sebaliknya, filter sharpening bertindak sebagai High-Pass Filter yang justru mencari dan memperkuat komponen frekuensi tinggi tersebut.
Filter ini meningkatkan kontras lokal pada area yang memiliki perubahan gradien warna yang tajam. Dampaknya pada detail gambar sangat
signifikan; guratan tekstur pada latar belakang akan terlihat jauh lebih tegas, kasar, dan menonjol, sementara batas kontur objek utama akan 
tampak lebih tajam (sharp edges). Kendati demikian, penajaman yang agresif memiliki efek samping, yaitu turut memperkuat dan memperjelas noise atau 
bintik halus yang ada pada gambar asli, sehingga berisiko membuat gambar tampak kotor (grainy) atau memicu munculnya efek halo di sekitar tepi objek.
````
